```python
# Chuyển danh sách cần tìm và cột labname về dạng chữ thường để so sánh chuẩn xác
lab_name_need_lower = [name.lower() for name in ["Creatinine", "BUN", "Lactate", "WBC", "Hemoglobin", "Platelets", "Bilirubin", "Glucose"]]

# Lọc và đếm số lượng
counts = lab_df[lab_df["labname"].str.lower().isin(lab_name_need_lower)]["labname"].value_counts()
print(counts)
```

```python
# 1. Chuyển đổi tên xét nghiệm về chữ thường và cắt bỏ khoảng trắng thừa
lab_df["labname_clean"] = lab_df["labname"].astype(str).str.lower().str.strip()

# 2. Thống kê và in ra 30 loại xét nghiệm xuất hiện nhiều nhất trong bảng
top_labs = lab_df["labname_clean"].value_counts().head(30)
print(top_labs)
```

```python
# 1. Định nghĩa từ điển chuẩn hóa tên xét nghiệm (Mapping Dictionary)
lab_mapping = {
    "glucose": "glucose",
    "bedside glucose": "glucose",      # Gộp chung vào glucose
    "bun": "bun",
    "creatinine": "creatinine",
    "lactate": "lactate",
    "wbc x 1000": "wbc",               # Đổi tên gọn lại thành wbc
    "platelets x 1000": "platelets",   # Đổi tên gọn lại thành platelets
    "hgb": "hemoglobin",
    "total bilirubin": "bilirubin",
}

# 2. Tạo cột lab_mapped dựa trên từ điển trên (những cái không có trong từ điển sẽ biến thành NaN hoặc giữ nguyên tùy ý)
lab_df["lab_mapped"] = lab_df["labname_clean"].map(lab_mapping)

# 3. Lọc bỏ các dòng lỗi (#name?) và các xét nghiệm không nằm trong danh sách quan tâm
lab_clean_df = lab_df.dropna(subset=["lab_mapped"])

print("Số lượng sau khi chuẩn hóa và lọc:")
print(lab_clean_df["lab_mapped"].value_counts())
```
